```python
SELECT 
*
FROM OPENROWSET(
                BULK 'https://advprojectst.dfs.core.windows.net/silver/AdventureWorks_Calendar/',
                FORMAT = 'PARQUET'
) as quary1
--------------------------
CREATE SCHEMA gold;

----- create view on top of the quary---

CREATE VIEW gold.Calendar
AS
SELECT
* 
FROM 
OPENROWSET (
    BULK 'https://advprojectst.dfs.core.windows.net/silver/AdventureWorks_Calendar/',
    FORMAT = 'PARQUET'
) AS QUARY1

------ CREATE VIEW ON Customers----
CREATE VIEW gold.Customers
AS
SELECT
*
FROM
OPENROWSET (
    BULK 'https://advprojectst.dfs.core.windows.net/silver/AdventureWorks_Customers/',
     FORMAT = 'PARQUET'
) AS QUARY1

-- CREATE VIEW PRODUCTS
------------------------
CREATE VIEW gold.products
AS
SELECT 
    * 
FROM 
    OPENROWSET
        (
            BULK 'https://advprojectst.dfs.core.windows.net/silver/AdventureWorks_Products/',
            FORMAT = 'PARQUET'
        ) as QUARY1
DROP VIEW gold.products;
        -- CREATE VIEW RETURNS
------------------------
CREATE VIEW gold.returns
AS
SELECT 
    * 
FROM 
    OPENROWSET
        (
            BULK 'https://advprojectst.dfs.core.windows.net/silver/AdventureWorks_Returns/',
            FORMAT = 'PARQUET'
        ) as QUARY1
        
        -- CREATE VIEW SALES
------------------------
CREATE VIEW gold.sales
AS
SELECT 
    * 
FROM 
    OPENROWSET
        (
            BULK 'https://advprojectst.dfs.core.windows.net/silver/AdventureWorks_Sales/',
            FORMAT = 'PARQUET'
        ) as QUARY1

------------------------
-- CREATE VIEW TERRITORIES
------------------------
CREATE VIEW gold.territories
AS
SELECT 
    * 
FROM 
    OPENROWSET
        (
            BULK 'https://advprojectst..core.windows.net/silver/AdventureWorks_Territories/',
            FORMAT = 'PARQUET'
        ) as QUER1
DROP VIEW gold.territories;
CREATE MASTER KEY ENCRYPTION BY PASSWORD = 'GmailGmail@2026'

------ CREATE CREDENTICAL

CREATE DATABASE SCOPED CREDENTIAL cred_adv
 WITH
     IDENTITY = 'Managed Identity'

---------- createing external data source 

CREATE EXTERNAL DATA SOURCE source_silver
WITH (
		LOCATION = 'https://advprojectst.dfs.core.windows.net/silver',

		CREDENTIAL = cred_adv

)

CREATE EXTERNAL DATA SOURCE source_gold
WITH (
		LOCATION = 'https://advprojectst.dfs.core.windows.net/gold',

		CREDENTIAL = cred_adv

)


-----creating external file format-----

CREATE EXTERNAL FILE FORMAT format_parquet
WITH (
    FORMAT_TYPE = PARQUET,
    DATA_COMPRESSION = 'org.apache.hadoop.io.compress.SnappyCodec'
)

----createing external tables extsales-----

CREATE EXTERNAL TABLE gold.extsales
WITH
(
	LOCATION = 'extsales',
	DATA_SOURCE = source_gold,
	FILE_FORMAT = format_parquet
)
AS
SELECT *
FROM gold.sales

SELECT * FROM gold.extsales;

------ CREATE EXTERNAL TABLE ON CUSTOMERS ------

CREATE EXTERNAL TABLE gold.extcustomers
WITH (
    LOCATION = 'extcustomers',
    DATA_SOURCE = source_gold,
    FILE_FORMAT = format_parquet
)
AS
SELECT *
FROM gold.customers;


------ CREATE EXTERNAL TABLE ON PRODUCTS ------

CREATE EXTERNAL TABLE gold.extproducts
WITH (
    LOCATION = 'extproducts',
    DATA_SOURCE = source_gold,
    FILE_FORMAT = format_parquet
)
AS
SELECT *
FROM gold.products;


------ CREATE EXTERNAL TABLE ON CALENDAR ------

CREATE EXTERNAL TABLE gold.extcalendar
WITH (
    LOCATION = 'extcalendar',
    DATA_SOURCE = source_gold,
    FILE_FORMAT = format_parquet
)
AS
SELECT *
FROM gold.calendar;


------ CREATE EXTERNAL TABLE ON RETURNS ------

CREATE EXTERNAL TABLE gold.extreturns
WITH (
    LOCATION = 'extreturns',
    DATA_SOURCE = source_gold,
    FILE_FORMAT = format_parquet
)
AS
SELECT *
FROM gold.returns;


------ CREATE EXTERNAL TABLE ON TERRITORIES ------

CREATE EXTERNAL TABLE gold.extterritories
WITH (
    LOCATION = 'extterritories',
    DATA_SOURCE = source_gold,
    FILE_FORMAT = format_parquet
)
AS
SELECT *
FROM gold.territories;

```
